### D11. `while True` and `break`: add numbers until the user types `done`

1. Create a variable `total = 0`.
2. Use a `while True:` loop. Inside it, ask: `Enter a number (or done to finish): `
3. If the user types `done`, stop the loop with `break`.
4. Otherwise, convert what they typed with `float()` and add it to `total`.
5. After the loop, print the total.

**Example run:**

```
Enter a number (or done to finish): 5
Enter a number (or done to finish): 10
Enter a number (or done to finish): 2.5
Enter a number (or done to finish): done
Total: 17.5
```

*In a notebook, the input box appears at the top of VS Code. Press Enter after each number.*

````python
#Create a variable
total = 0
#loop around while the condition is true
while True:
    #Prompt a user to enter a number or done to finish
    number = input("Enter a number (or done to finish): ")
    #check if  the user input is corresponding to done
    if number == "done":
        #Display the total that a user has entered the numbers
        print(f"Total: {total}")
        #break the loop condition
        break
    else:
        #Add number to the total
        total += float(number)
````

### D12. `continue`: skip one number

**Expected output:** `1 2 3 4 6 7 8 9 10` (one per line)

*Hint: add 1 to your counter **before** the `continue`, or the loop gets stuck on 5.*

````python
# Create a variable and assign a value to it
count = 0
#loop to check the condition if count is less than or equal to 9
while count <= 9:
    #Increment count by  1
    count += 1
    #check the condition if the value of count is corresponding to 5
    if count == 5:
        #Skip the counter of 5
        continue
    #Display the count 
    print(count)
````

### D13. `range(stop)` and `range(start, stop)`

1. Use `for i in range(5):` to print the numbers it gives you.
2. Then write a second loop using `range(start, stop)` that prints 1 to 5.

**Expected output:** `0 1 2 3 4`, then `1 2 3 4 5` (one per line)

*Remember: the `stop` number is never included.*

````python
#loop around for each index with in range 5
for i in range(5):
    #Display the index
    print(i)
#loop around for each index starting by 1 and stopping with in range 6
for i in range(1, 6):
    print(i)
````

### D14. `range(start, stop, step)`

1. Print every even number from 2 to 20.
2. Then write a second loop that counts down from 10 to 1.

**Expected output:** `2 4 6 … 20`, then `10 9 8 … 1` (one per line)

*Hint: to count down, use a negative step.*

````python
#loop around for each index starting by 2, stepping in by 2 and stopping with  in range 22
for i in range(2, 22, 2):
    #DIsplay the index
    print(i)
#loop around for each count starting by 10, stepping in by -1 and stopping with in range 0
for count in range(10, 0, -1):
    print(count)
````

### D15. `enumerate()`: the position and the value together

1. Create `code = "A7X"`.
2. Use a `for` loop with `enumerate(code, start=1)` to print each character with its position.

**Expected output:**
```
Character 1: A
Character 2: 7
Character 3: X
```

````python
#Create a variable and assign a string to it
code = "A7X"
#loop around for each index and char while using enumurate 
for i, char in enumerate(code, start=1):
    #Display the character with the index and the char
    print(f"Character {i}: {char}")
 
````

### C1. Check three records in one run

Write a `for` loop that runs exactly 3 times. Each time, ask for a `value` and a `limit`,
and print `OVER LIMIT` if `value` is greater than `limit`, otherwise `OK`.

**Test it** with these three pairs, in this order:

| value | limit | Expected output |
|---|---|---|
| 87 | 100 | `OK` |
| 120 | 100 | `OVER LIMIT` |
| 45 | 100 | `OK` |

````python
#Creating an open interface
print("-" * 30)
#Display the label
print(" Value  limit  Expected output")
#loop around for each count with in range 3
for count in range(3):
    #Create a variable and prompt a user to enter a value 
    value = float(input("Please enter a first value: "))
    #Create a variable and prompt a user to enter a limit
    limit = float(input("Please enter a second value: "))
    #Creating another interface
    print("-" * 30)
    #Check the condition if value is greater than limit
    if value > limit:
        #Display the value, limit and over limit if the condition is true
        print(f" {value:<1.0f}    {limit:.0f}    OVER LIMIT")
    else:
        #Display the value, limit and Ok if the condition is false
        print(f" {value:<1.0f}     {limit:.0f}    OK")
#Closing an interface
print("-" * 30)
````

### C2. Three tries to log in

The correct password is `python123`. The user gets **at most 3 tries**.

- If they type the correct password, print `Access granted` and stop asking.
- If they get it wrong 3 times, print `Account locked`.

**Test it twice:**

| You type | Expected output |
|---|---|
| `abc`, then `python123` | `Access granted` (after 2 tries) |
| `abc`, `123`, `pass` | `Account locked` |

````python
#Create a variable and assign a string to it 
code = "python123"
#Create a variable and assign a value to it
login_attempts = 0
#Create a variable and assign a boolean to it
Account_Status = "False"
#loop if the Account_Status is false
while Account_Status == "False":
    #Create a variable and prompt a user to enter your password
    password = input("Please enter your password: ")
    #Check the password that a user has entered is corresponding to the code
    if password == code:
        #Display Access granted if a user get the password correct
        print("Access granted")
        #Set Account_Status to true to go out while loop
        Account_Status = "True"
        #Check the number of login attempts that a user has entered the password if the login attempts are more than or equal 2
    elif login_attempts >= 2:
        #Display Account locked if a user has failed to login for more than or equal to 2
        print("Account locked")
        #Set Account_Status to true to go out while loop
        Account_Status = "True"
        #Check if the two conditions above that are false
    else:
        #Display a message that a user does not get password correct
        print("Invalid Password")
        #Increment login attempts by 1
        login_attempts += 1
````
