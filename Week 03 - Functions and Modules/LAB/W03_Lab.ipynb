### D8. `*args` and `**kwargs`

a) Define `total(*numbers)` that returns the sum of all the numbers passed in. Print `total(1, 2, 3)`,
   `total(10, 20, 30, 40)` and `total()`.

b) Define `configure(**settings)` that loops over `settings.items()` and prints each one as `key = value`.
   Call it with `configure(host="10.0.0.1", port=8080)`.

**Expected output:**
```
6
100
0
host = 10.0.0.1
port = 8080
```

````python
#a)
#Define a function which is called total(*numbers)
def total(*numbers):
    """Do the sum of all numbers that are inside the bracket"""
    #Create a variable and intialise it
    sum = 0
    #loop around the numbers to seperate them
    for num in numbers:
        #Sum is incremented by num
        sum += num
    #Return the value of sum to a user
    return sum
#Call the function of total(*numbers) that contains three parameters inside the bracket and display the sum of three numbers for the first time    
print(total(1, 2, 3))
#Call the function of total(*numbers) that contains four parameters inside the bracket and display the sum of four numbers for the second time 
print(total(10, 20, 30, 40))
#Call the function of total(*numbers) that contains zero parameter inside the bracket and display the sum of zero number for the third time
print(total())
#b)
#Define a function which is called configure(**settings)
def configure(**settings):
    """Display the values based assignment of variable defined below"""
    #loop around the settings item to seperate them
    for key, value in settings.items():
        #Display the assignment of key to value 
        print(key,"=", value)
#Call the function of configure(**settings) that contains two parameters inside the bracket
configure(host="10.0.0.1", port=8080)        
````

### C2. How many are over?

Write `count_over(*values, limit=100)` with a one-line docstring. It returns how many of the values
are over the limit.

| Call | Expected output |
|---|---|
| `count_over(87, 120, 45, 101)` | `2` |
| `count_over(87, 120, 45, 101, limit=50)` | `3` |
| `count_over()` | `0` |

````python
#Define a function which is called count_over(*values, limit=100)
def count_over(*values, limit=100):
    """Checking if a value of the limit of 100, display the count of occurrences that are over this limit."""
    #Create a variable and initialise it
    count_limit = 0
    #loop around the values to seperate them
    for value in values:
        #Check the condition if value is greater than or equal to 100
        if value >= limit:
            #count_limit is incremented by 1
            count_limit += 1
    #Return the count_limit to a user
    return count_limit
#Call a function which contains 4 parameters and display it for the first time
print(count_over(87, 120, 45, 101))
#Call a function which contains 5 parameters and display it for the second time
print(count_over(97, 120, 45, 101, limit=50))
#Call a function which contains 0 parameter and display it for the third time
print(count_over())
````
